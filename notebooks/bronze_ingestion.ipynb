```python
from pyspark.sql.functions import current_timestamp, input_file_name
```

```python
deliveries_df = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferschema", "true")\
    .load("Files/deliveries.csv")

display(deliveries_df)
```

```python
deliveries_bronze = deliveries_df \
    .withColumn("ingestion_timestamp", current_timestamp())\
    .withColumn("source_file_name", input_file_name())
```

```python
deliveries_bronze.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_deliveries")
```

```python
matches_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("Files/matches.csv")
```

```python
display(matches_df)
```

```python
matches_bronze=matches_df \
    .withColumn("ingestion_timestamp", current_timestamp())\
    .withColumn("source_file_name", input_file_name())

matches_bronze.write.format("delta") \
    .mode("overwrite")\
    .saveAsTable("bronze_matches")
```
